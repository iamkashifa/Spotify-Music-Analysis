```python
import seaborn as sns
fig, axes = plt.subplots(1, 2, figsize=(18, 7))
fig.patch.set_facecolor('#f0f0f0') #background color for the figure

#Positive Correlation Matrix

# Define the features for the first heatmap as seen in the image.
features_positive = ['valence', 'loudness', 'danceability', 'energy', 'time_signature']

# Check if all required columns exist in the DataFrame
if all(feature in df.columns for feature in features_positive):

    # Calculate the correlation matrix for the selected features.
    corr_matrix_pos = df[features_positive].corr()

    # Generate the heatmap.
    # annot=True displays the correlation values on the map.
    # cmap='Greens' sets the color scheme.
    # fmt='.2g' formats the numbers to two significant figures.
    sns.heatmap(
        corr_matrix_pos,
        ax=axes[0],
        annot=True,
        cmap='Greens',
        fmt='.2g',
        linewidths=.5
    )

    # Set the title for the first subplot.
    axes[0].set_title('Correlation Matrix of Selected Spotify Features', fontsize=14, pad=20)
else:
    axes[0].text(0.5, 0.5, 'Could not generate positive correlation plot.\nOne or more columns not found in the data.',
                 ha='center', va='center', fontsize=12, color='red')
    axes[0].set_title('Correlation Matrix of Selected Spotify Features', fontsize=14, pad=20)
    #Negative Correlations Matrix ---

# Define the features for the second heatmap.
features_negative = ['energy', 'acousticness', 'loudness', 'instrumentalness', 'time_signature', 'danceability']

# Check if all required columns exist in the DataFrame
if all(feature in df.columns for feature in features_negative):

    # Calculate the correlation matrix for these features.
    corr_matrix_neg = df[features_negative].corr()

    # Create a mask to hide positive correlations and the diagonal.
    # We want to see only the cells with negative correlation values.
    # The mask is True for values we want to hide.
    mask_neg = corr_matrix_neg >= 0

    # Generate the second heatmap.
    # We use the mask to hide the positive values.
    # The colormap 'Reds' is used for negative correlations. We reverse it to go from light to dark.
    # vmin and vmax are set to focus the color scale on the negative range.
    sns.heatmap(
        corr_matrix_neg,
        ax=axes[1],
        annot=True,
        cmap='Reds_r', # A reversed Reds colormap (light red to dark red)
        mask=mask_neg,
        fmt='.2g',
        linewidths=.5,
        vmin=-1, # Set the minimum for the color bar
        vmax=0   # Set the maximum for the color bar
    )

    # Set the title for the second subplot.
    axes[1].set_title('Negative Correlations Matrix of Selected Spotify Features (Range: -1 to 0)', fontsize=14, pad=20)
else:
    axes[1].text(0.5, 0.5, 'Could not generate negative correlation plot.\nOne or more columns not found in the data.',
                 ha='center', va='center', fontsize=12, color='red')
    axes[1].set_title('Negative Correlations Matrix of Selected Spotify Features (Range: -1 to 0)', fontsize=14, pad=20)


#Final Touches and Display

# Use tight_layout to ensure plots and titles don't overlap.
plt.tight_layout(pad=3.0)

# Display the plots.
plt.show()
```

```python

# Find the top 5 unique popularity scores from the dataset.
top_popularity_scores = df['track_popularity'].value_counts().nlargest(5).index

# Filter the DataFrame to only include tracks with those top popularity scores.
top_tracks_df = df[df['track_popularity'].isin(top_popularity_scores)]


#Prepare for Plotting
# Create a figure and a set of subplots for the two bar charts.
fig, axes = plt.subplots(1, 2, figsize=(16, 6))
fig.patch.set_facecolor('#ffffff') # Set a white background for the figure


#Livenes

# Check if the required columns exist
if 'liveness' in top_tracks_df.columns and 'track_popularity' in top_tracks_df.columns:
    # Create the bar plot for liveness. Seaborn automatically calculates the mean for each category.
    sns.barplot(
        x='track_popularity',
        y='liveness',
        data=top_tracks_df,
        ax=axes[0],
        palette='viridis_r', # Use a reversed viridis color palette
        order=sorted(top_popularity_scores) # Ensure the bars are ordered by popularity
    )

    # Set titles and labels
    axes[0].set_title('Liveness of Top 10 Most Popular Tracks', fontsize=14, pad=15)
    axes[0].set_ylabel('Liveness', fontsize=12)
    axes[0].set_xlabel('Track Popularity', fontsize=12)
else:
    axes[0].text(0.5, 0.5, "Required columns ('liveness', 'track_popularity') not found.", ha='center', va='center', color='red')
    axes[0].set_title('Liveness of Top 10 Most Popular Tracks', fontsize=14, pad=15)


#Loudness

# Check if the required columns exist
if 'loudness' in top_tracks_df.columns and 'track_popularity' in top_tracks_df.columns:
    # Create the bar plot for loudness
    sns.barplot(
        x='track_popularity',
        y='loudness',
        data=top_tracks_df,
        ax=axes[1],
        palette='viridis_r', # Use the same palette for consistency
        order=sorted(top_popularity_scores) # Order the bars
    )

    # Set titles and labels
    axes[1].set_title('Loudness of Top 10 Most Popular Tracks', fontsize=14, pad=15)
    axes[1].set_ylabel('Loudness', fontsize=12)
    axes[1].set_xlabel('Track Popularity', fontsize=12)
else:
    axes[1].text(0.5, 0.5, "Required columns ('loudness', 'track_popularity') not found.", ha='center', va='center', color='red')
    axes[1].set_title('Loudness of Top 10 Most Popular Tracks', fontsize=14, pad=15)


#Final Touches and Display

# Adjust layout to prevent labels from overlapping
plt.tight_layout(pad=3.0)

# Show the final plot
plt.show()
```

```python

# Define the features we want to correlate with track popularity.
features = [
    'track_popularity', 'valence', 'tempo', 'energy',
    'danceability', 'loudness', 'liveness', 'speechiness'
]

# Check if all required columns exist in the DataFrame
if all(feature in df.columns for feature in features):
    # Calculate the full correlation matrix for the selected features.
    corr_matrix = df[features].corr()

    # Isolate the correlations with 'track_popularity'.
    # We convert it to a DataFrame for compatibility with seaborn's heatmap function.
    popularity_corr = corr_matrix[['track_popularity']].sort_values(by='track_popularity', ascending=False)


    # Create a figure and axes for the plot.
    plt.figure(figsize=(6, 8))

    # Generate the heatmap.
    # annot=True displays the correlation values.
    # cmap='coolwarm' provides a blue-to-red color scale.
    # vmin and vmax ensure the color scale is centered around 0.
    sns.heatmap(
        popularity_corr,
        annot=True,
        cmap='coolwarm',
        fmt='.3g', # Format numbers to 3 significant figures
        linewidths=.5,
        vmin=-1,
        vmax=1
    )

    #Final Touches and Display
    # Set the title for the plot.
    plt.title('Correlation of Features with Track Popularity', fontsize=14, pad=15)

    # Adjust layout to ensure everything fits.
    plt.tight_layout()

    # Display the plot.
    plt.show()

else:
    # If columns are missing, print an error message.
    print("Could not generate the plot because one or more required columns are missing from the data.")
    print("Required columns are:", features)
```

```python
import pandas as pd
import matplotlib.pyplot as plt



# Ensure the 'track_album_release_date' column is of string type
df['track_album_release_date'] = df['track_album_release_date'].astype(str)

# Convert the 'track_album_release_date' column to datetime
df['track_album_release_date'] = pd.to_datetime(df['track_album_release_date'], errors='coerce')

# Create a scatter plot to visualize the relationship
plt.figure(figsize=(12, 6))
plt.scatter(df['track_album_release_date'], df['track_popularity'], color='green', alpha=0.6)

# Adding titles and labels
plt.xlabel('Release Date')
plt.ylabel('Track Popularity')
plt.title('Relationship between Release Date and Track Popularity')
plt.grid(True)

# Show the plot
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


# Filter the most popular playlists
most_popular_playlists = df.groupby('playlist_name')['track_popularity'].mean().sort_values(ascending=False).head(3)
popular_playlists_df = df[df['playlist_name'].isin(most_popular_playlists.index)]

# Calculate the average valence, liveness, and acousticness for each popular playlist
playlist_stats = popular_playlists_df.groupby('playlist_name')[['valence', 'liveness', 'acousticness']].mean().reset_index()

# Display the grouped data
print("Key Differences between the Most Popular Playlists:")
print(playlist_stats)

# Create bar plots to visualize the differences
fig, axs = plt.subplots(1, 3, figsize=(18, 6))

sns.barplot(x='playlist_name', y='valence', data=playlist_stats, palette='viridis', ax=axs[0])
axs[0].set_title('Average Valence')
axs[0].set_xlabel('Playlist Name')
axs[0].set_ylabel('Valence')

sns.barplot(x='playlist_name', y='liveness', data=playlist_stats, palette='viridis', ax=axs[1])
axs[1].set_title('Average Liveness')
axs[1].set_xlabel('Playlist Name')
axs[1].set_ylabel('Liveness')

sns.barplot(x='playlist_name', y='acousticness', data=playlist_stats, palette='viridis', ax=axs[2])
axs[2].set_title('Average Acousticness')
axs[2].set_xlabel('Playlist Name')
axs[2].set_ylabel('Acousticness')

plt.suptitle('Differences in Valence, Liveness, and Acousticness across Most Popular Playlists')
plt.tight_layout()
plt.show()
```
